```python

name = input("Enter your Name: ")

result = "Hello " + name + ", welcome to Python programming"

print(result)
 
```

```python
# print first character 
print(result) 
print(result[0])
# last character
print(result[-1]) 
# first 5 characters of string
print(result[:5]) 
# last 11 characters of string
print(result[-11:]) 
# reverse 
print(result[::-1]) 
# using Slicing 
start = result.index("Python")
print("Word Python:", result[start:start + 6])
```

# String Methods

```python
strM=("Python beginner tutorial")
print(strM) 
# UPPERCASE 
print(strM.upper())
# LOWERCASE 
print(strM.lower()) 
# CAPITALISE 
print(strM.capitalize()) 
# count the occurrences of character ‘t’  in the string 
strM.count("t") 
# Replace Python with Machine Learning
strM.replace("Python", "Machine Learning")
```

# TUPLE

```python
#Concatenate the two tuples and store it in “t_combine” 
t1=('10','20','30')
t2=('40','50','60') 
t_combine=t1+t2
print(t_combine) 
#  Repeat the elements of “t_combine” 3 times 
print(t_combine*3) 
# Access the 3rd element from “t_combine” 
print(t_combine[2]) 
# Access the first three elements from “t_combine”
print(t_combine[:3]) 
#  Access the last three elements from “t_combine” 
print(t_combine[-3:])
```
